```python
import pandas as pd
train = pd.read_csv("housing_raw.csv")
train.head()
```

```python
# Menampilkan statistik deskriptif dari dataset
train.describe(include="all")
```

```python
# Memeriksa jumlah nilai yang hilang di setiap kolom
missing_values = train.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Mengatasi missing value
less = missing_values[missing_values < 1000].index
over = missing_values[missing_values >= 1000].index
```

```python
# Contoh mengisi nilai yang hilang dengan median untuk kolom numerik
numeric_features = train[less].select_dtypes(include=['number']).columns
train[numeric_features] = train[numeric_features].fillna(train[numeric_features].median())
```

```python
# Contoh mengisi nilai yang hilang dengan mode untuk kolom kategori
kategorical_features = train[less].select_dtypes(include=['str']).columns

for column in kategorical_features:
    train[column] = train[column].fillna(train[column].mode()[0])
```

```python
# Menghapus kolom dengan terlalu banyak nilai yang hilang
df = train.drop(columns=over)
```

```python
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Mengatasi Outliers
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot of {feature}')
    plt.show()
```

```python
# Contoh sederhana untuk mengidentifikasi outliers menggunakan IQR
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1
```

```python
# Filter dataframe untuk hanya menyimpan baris yang tidak mengandung outliers pada kolom numerik
condition = ~(
    (df[numeric_features] < (Q1 - 1.5 * IQR)) |
    (df[numeric_features] > (Q3 + 1.5 * IQR))
).any(axis=1)

df_filtered_numeric = df.loc[condition, numeric_features]
  
# Menggabungkan kembali dengan kolom kategorikal
categorical_features = df.select_dtypes(include=['str']).columns

df = pd.concat(
    [df_filtered_numeric, df.loc[condition, categorical_features]],
    axis=1
)
```

```python
# Mengganti outlier dengan nilai median
for column in numeric_features:
    Q1_col = df[column].quantile(0.25)
    Q3_col = df[column].quantile(0.75)
    IQR_col = Q3_col - Q1_col

    lower_bound = Q1_col - 1.5 * IQR_col
    upper_bound = Q3_col + 1.5 * IQR_col

    median = df[column].median()

    df[column] = df[column].apply(
        lambda x: median if x < lower_bound or x > upper_bound else x
    )
```

```python
# Mengganti outlier dengan nilai batas terdekat
for column in numeric_features:
    Q1_col = df[column].quantile(0.25)
    Q3_col = df[column].quantile(0.75)
    IQR_col = Q3_col - Q1_col

    lower_bound = Q1_col - 1.5 * IQR_col
    upper_bound = Q3_col + 1.5 * IQR_col

    df[column] = df[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )
```

```python
from sklearn.preprocessing import StandardScaler

# Standardisasi fitur numerik
scaler = StandardScaler()
df[numeric_features] = scaler.fit_transform(df[numeric_features])
```

## Evaluasi Model

Ketiga model regresi dievaluasi pada data uji (`x_test`, `y_test`) dengan tiga metrik: MAE (*Mean Absolute Error*), MSE (*Mean Squared Error*), dan R².

### 1. MAE (Mean Absolute Error)

$$MAE = \frac{1}{n}\sum_{i=1}^{n}\left|y_i - \hat{y}_i\right|$$

- Rata-rata selisih mutlak antara nilai asli ($y_i$) dan nilai prediksi ($\hat{y}_i$).
- Satuannya **sama dengan target**, yaitu `median_house_value`, sehingga mudah dibaca: MAE = 40.000 berarti prediksi rata-rata meleset sekitar 40.000 dari harga sebenarnya.
- Semua kesalahan diperlakukan sama, jadi tidak terlalu terpengaruh oleh outlier.
- **Semakin kecil semakin baik** (0 = prediksi sempurna).

### 2. MSE (Mean Squared Error)

$$MSE = \frac{1}{n}\sum_{i=1}^{n}\left(y_i - \hat{y}_i\right)^2$$

- Rata-rata dari **kuadrat** selisih antara nilai asli dan prediksi.
- Karena dikuadratkan, kesalahan besar dihukum jauh lebih berat daripada kesalahan kecil, sehingga MSE sensitif terhadap outlier.
- Satuannya adalah satuan target **kuadrat**, jadi angkanya sangat besar dan sulit dibaca langsung. Untuk interpretasi, ambil akarnya: $RMSE = \sqrt{MSE}$, yang satuannya kembali sama dengan target.
- **Semakin kecil semakin baik.**

### 3. R² (Koefisien Determinasi)

$$R^2 = 1 - \frac{\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}{\sum_{i=1}^{n}(y_i - \bar{y})^2}$$

- Menunjukkan **seberapa besar variasi target yang bisa dijelaskan oleh model**, dibandingkan dengan model sederhana yang selalu menebak rata-rata ($\bar{y}$).
- Tidak punya satuan, umumnya bernilai 0 sampai 1. R² = 0,76 berarti sekitar 76% variasi harga rumah dapat dijelaskan model.
- R² = 1 berarti sempurna, R² mendekati 0 berarti model tidak lebih baik dari menebak rata-rata, dan R² negatif berarti model lebih buruk dari menebak rata-rata.
- **Semakin besar (mendekati 1) semakin baik.**

### Ringkasan

| Metrik | Mengukur | Satuan | Arah yang baik |
|--------|----------|--------|----------------|
| MAE | Rata-rata besar kesalahan | Sama dengan target | Makin kecil |
| MSE | Rata-rata kuadrat kesalahan (peka outlier) | Target kuadrat | Makin kecil |
| R² | Proporsi variasi yang dijelaskan model | Tanpa satuan | Makin dekat 1 |

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Evaluasi pada model LARS
pred_lars = lars.predict(x_test)
mae_lars = mean_absolute_error(y_test, pred_lars)
mse_lars = mean_squared_error(y_test, pred_lars)
r2_lars = r2_score(y_test, pred_lars)

# Membuat dictionary untuk menyimpan hasil evaluasi
data = {
    'MAE': [mae_lars], 
    'MSE': [mse_lars], 
    'R2': [r2_lars]
}

# Konversi dictionary menjadi DataFrame
df_results = pd.DataFrame(data, index=['Lars'])
df_results
```

```python
# Evaluasi pada model Linear Regression
pred_LR = LR.predict(x_test)
mae_LR = mean_absolute_error(y_test, pred_LR)
mse_LR = mean_squared_error(y_test, pred_LR)
r2_LR = r2_score(y_test, pred_LR)

# Menambahkan hasil evaluasi LR ke DataFrame
df_results.loc['Linear Regression'] = [mae_LR, mse_LR, r2_LR]
df_results
```

```python
# Evaluasi pada model Gradient Boosting Regressor
pred_GBR = GBR.predict(x_test)
mae_GBR = mean_absolute_error(y_test, pred_GBR)
mse_GBR = mean_squared_error(y_test, pred_GBR)
r2_GBR = r2_score(y_test, pred_GBR)

# Menambahkan hasil evaluasi GBR ke DataFrame
df_results.loc['GradientBoostingRegressor'] = [mae_GBR, mse_GBR, r2_GBR]
df_results
```

### Interpretasi Hasil Evaluasi

| Model | MAE | MSE | RMSE (akar MSE) | R² |
|-------|-----|-----|-----------------|----|
| Lars | 89.963 | 1,297 × 10¹⁰ | ≈ 113.902 | 0,015 |
| Linear Regression | 67.958 | 8,085 × 10⁹ | ≈ 89.918 | 0,386 |
| Gradient Boosting Regressor | 39.299 | 3,155 × 10⁹ | ≈ 56.166 | 0,761 |

- **Lars** paling lemah. R² ≈ 0,015 berarti model nyaris tidak menjelaskan variasi harga rumah, wajar karena `n_nonzero_coefs=1` membatasi model hanya memakai satu fitur.
- **Linear Regression** sudah lebih baik (R² ≈ 0,39), tetapi hubungan antara fitur dan harga rumah kemungkinan tidak sepenuhnya linear sehingga kesalahannya masih besar.
- **Gradient Boosting Regressor** terbaik di ketiga metrik: MAE dan MSE paling kecil, R² paling tinggi (≈ 0,76). Rata-rata prediksinya meleset sekitar 39 ribu, dan model menjelaskan sekitar 76% variasi harga.
- RMSE jauh lebih besar daripada MAE pada semua model, menandakan ada sejumlah prediksi dengan kesalahan besar (kemungkinan rumah berharga sangat tinggi atau sangat rendah) yang ikut menaikkan MSE.

**Kesimpulan:** Gradient Boosting Regressor dipilih sebagai model terbaik, dan model inilah yang disimpan pada tahap berikutnya.

```python
%%writefile ml_utils.py
"""Fungsi bersama untuk API dan monitoring: memuat bundle, preprocessing, prediksi, logging."""
import json
from pathlib import Path

import joblib
import pandas as pd

BUNDLE_PATH = "model_bundle.joblib"


def load_bundle(path=BUNDLE_PATH):
    # Peringatan: joblib/pickle bisa mengeksekusi kode saat dimuat.
    # Hanya muat file model yang kamu buat sendiri atau dari sumber tepercaya.
    return joblib.load(path)


def preprocess(records, bundle):
    """List of dict (data mentah) -> array numerik siap masuk model."""
    df = pd.DataFrame(records)
    for col, enc in bundle["encoders"].items():
        unknown = set(df[col].dropna().unique()) - set(enc.classes_)
        if unknown or df[col].isna().any():
            raise ValueError(
                f"Nilai '{col}' tidak dikenal: {sorted(map(str, unknown)) or 'kosong'}. "
                f"Nilai yang valid: {list(enc.classes_)}"
            )
        df[col] = enc.transform(df[col])
    df = df.reindex(columns=bundle["feature_names"])
    return bundle["imputer"].transform(df.to_numpy(dtype=float))


def predict(records, bundle):
    return bundle["model"].predict(preprocess(records, bundle))


def append_jsonl(path, obj):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("a", encoding="utf-8") as f:
        f.write(json.dumps(obj, ensure_ascii=False,
                           default=lambda o: o.item() if hasattr(o, "item") else str(o)) + "\n")
```

```python
%%writefile app.py
"""API prediksi harga rumah (Flask).
Dev      : python app.py
Produksi : gunicorn -w 1 --threads 4 -b 0.0.0.0:8000 app:app
"""
import math
import os
import threading
import time
import uuid
from collections import deque
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
from flask import Flask, jsonify, request

from ml_utils import append_jsonl, load_bundle, predict

LOG_DIR = Path(os.getenv("LOG_DIR", "logs"))
PRED_LOG = LOG_DIR / "predictions.jsonl"
FEEDBACK_LOG = LOG_DIR / "feedback.jsonl"
MAX_ROWS = 1000

bundle = load_bundle()
app = Flask(__name__)

lock = threading.Lock()
stats = {"requests": 0, "errors": 0}
latencies = deque(maxlen=1000)

# nama fitur: (nilai minimum, nilai maksimum, boleh kosong). Sesuaikan jika kolom dataset berbeda.
FIELDS = {
    "longitude": (-125, -114, False),
    "latitude": (32, 43, False),
    "housing_median_age": (0, 100, False),
    "total_rooms": (1, None, False),
    "total_bedrooms": (0, None, True),    # boleh kosong -> diimputasi median training
    "population": (1, None, False),
    "households": (1, None, False),
    "median_income": (0.01, None, False),
}
CATEGORY = "ocean_proximity"
VALID_CATEGORIES = list(bundle["encoders"][CATEGORY].classes_)


def validate(record):
    clean, errors = {}, []
    for name, (lo, hi, optional) in FIELDS.items():
        v = record.get(name)
        if v is None:
            if optional:
                clean[name] = None
            else:
                errors.append(f"'{name}' wajib diisi")
            continue
        if isinstance(v, bool) or not isinstance(v, (int, float)) or not math.isfinite(v):
            errors.append(f"'{name}' harus berupa angka")
        elif (lo is not None and v < lo) or (hi is not None and v > hi):
            errors.append(f"'{name}' di luar rentang [{lo}, {hi}]")
        else:
            clean[name] = float(v)
    cat = record.get(CATEGORY)
    if cat not in VALID_CATEGORIES:
        errors.append(f"'{CATEGORY}' harus salah satu dari {VALID_CATEGORIES}")
    clean[CATEGORY] = cat
    return clean, errors


def count(error=False, latency_ms=None):
    with lock:
        stats["requests"] += 1
        stats["errors"] += int(error)
        if latency_ms is not None:
            latencies.append(latency_ms)


@app.get("/health")
def health():
    return jsonify(status="ok", model_version=bundle["model_version"])


@app.post("/predict")
def predict_endpoint():
    """Body: {"data": [ {fitur...}, ... ]}  atau satu objek fitur langsung."""
    start = time.perf_counter()
    payload = request.get_json(silent=True)
    if payload is None:
        count(error=True)
        return jsonify(error="Body harus berupa JSON"), 400

    items = payload.get("data", payload) if isinstance(payload, dict) else payload
    if isinstance(items, dict):
        items = [items]
    if not isinstance(items, list) or not items or not all(isinstance(i, dict) for i in items):
        count(error=True)
        return jsonify(error="'data' harus berisi satu atau lebih objek fitur"), 400
    if len(items) > MAX_ROWS:
        count(error=True)
        return jsonify(error=f"Maksimal {MAX_ROWS} baris per request"), 400

    cleaned, problems = [], []
    for idx, rec in enumerate(items):
        clean, errs = validate(rec)
        cleaned.append(clean)
        if errs:
            problems.append({"index": idx, "errors": errs})
    if problems:
        count(error=True)
        return jsonify(error="Input tidak valid", detail=problems), 422

    try:
        preds = predict(cleaned, bundle)
    except Exception:
        app.logger.exception("Prediksi gagal")
        count(error=True)
        return jsonify(error="Prediksi gagal"), 500

    latency_ms = (time.perf_counter() - start) * 1000
    count(latency_ms=latency_ms)
    now = datetime.now(timezone.utc).isoformat()
    ids = []
    for rec, pred in zip(cleaned, preds):
        rid = uuid.uuid4().hex
        ids.append(rid)
        append_jsonl(PRED_LOG, {
            "request_id": rid, "timestamp": now, "input": rec, "prediction": float(pred),
            "latency_ms": round(latency_ms, 2), "model_version": bundle["model_version"],
        })
    return jsonify(prediction=[round(float(p), 2) for p in preds], request_ids=ids,
                   model_version=bundle["model_version"])


@app.post("/feedback")
def feedback():
    """Kirim nilai aktual setelah diketahui: {"request_id": "...", "actual_value": 452600}"""
    body = request.get_json(silent=True) or {}
    rid, actual = body.get("request_id"), body.get("actual_value")
    if (not isinstance(rid, str) or isinstance(actual, bool)
            or not isinstance(actual, (int, float)) or not math.isfinite(actual) or actual <= 0):
        return jsonify(error="Wajib: request_id (teks) dan actual_value (angka > 0)"), 422
    append_jsonl(FEEDBACK_LOG, {"request_id": rid, "actual_value": float(actual),
                                "timestamp": datetime.now(timezone.utc).isoformat()})
    return jsonify(status="tersimpan")


@app.get("/metrics")
def metrics():
    with lock:
        lat = np.array(latencies) if latencies else np.array([0.0])
        n, err = stats["requests"], stats["errors"]
    return jsonify(requests=n, errors=err, error_rate=round(err / n, 4) if n else 0.0,
                   latency_ms_avg=round(float(lat.mean()), 2),
                   latency_ms_p95=round(float(np.percentile(lat, 95)), 2),
                   model_version=bundle["model_version"])


if __name__ == "__main__":
    app.run(host="0.0.0.0", port=8000, debug=False)
```

```python
%%writefile monitoring.py
"""Monitoring: kesehatan layanan, data drift, dan performa model. Bisa dijalankan terjadwal: python monitoring.py"""
import json
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.stats import ks_2samp
from sklearn.metrics import mean_absolute_error, r2_score

from ml_utils import load_bundle, preprocess

THRESHOLDS = {
    "psi_warn": 0.10,
    "psi_alert": 0.25,
    "mae_degradation": 0.20,   # MAE boleh naik maksimal 20% dari baseline
    "p95_latency_ms": 500,
    "min_samples": 100,        # di bawah ini hasil drift tidak andal
}


def read_jsonl(path):
    path = Path(path)
    if not path.exists():
        return pd.DataFrame()
    with path.open(encoding="utf-8") as f:
        return pd.DataFrame([json.loads(line) for line in f if line.strip()])


def psi(expected, actual):
    """Population Stability Index. <0.1 stabil, 0.1-0.25 waspada, >0.25 drift."""
    expected = np.asarray(expected, dtype=float)
    actual = np.asarray(actual, dtype=float)
    if len(np.unique(expected)) <= 10:          # fitur diskrit/kategori -> bandingkan proporsi
        cats = np.union1d(np.unique(expected), np.unique(actual))
        e = np.array([(expected == c).mean() for c in cats])
        a = np.array([(actual == c).mean() for c in cats])
    else:                                       # fitur kontinu -> bin berdasarkan kuantil data training
        edges = np.unique(np.quantile(expected, np.linspace(0, 1, 11)))
        edges[0], edges[-1] = -np.inf, np.inf
        e = np.histogram(expected, edges)[0] / len(expected)
        a = np.histogram(actual, edges)[0] / len(actual)
    e, a = np.clip(e, 1e-4, None), np.clip(a, 1e-4, None)
    return float(np.sum((a - e) * np.log(a / e)))


def drift_report(reference, live, features):
    rows = []
    for col in features:
        p = psi(reference[col], live[col])
        ks = ks_2samp(reference[col], live[col])
        status = ("DRIFT" if p >= THRESHOLDS["psi_alert"]
                  else "WASPADA" if p >= THRESHOLDS["psi_warn"] else "STABIL")
        rows.append({"fitur": col, "psi": round(p, 3), "ks_stat": round(float(ks.statistic), 3),
                     "ks_pvalue": round(float(ks.pvalue), 4), "status": status})
    return pd.DataFrame(rows).sort_values("psi", ascending=False).reset_index(drop=True)


def run_monitoring(log_dir="logs", bundle_path="model_bundle.joblib",
                   reference_path="reference_data.csv", save=True, verbose=True):
    bundle = load_bundle(bundle_path)
    features = bundle["feature_names"]
    preds = read_jsonl(Path(log_dir) / "predictions.jsonl")
    fb = read_jsonl(Path(log_dir) / "feedback.jsonl")
    alerts, result = [], {"n_predictions": len(preds), "n_feedback": len(fb)}

    if preds.empty:
        alerts.append("Belum ada log prediksi.")
        return {"alerts": alerts, **result}

    # 1) Layanan
    p95 = float(np.percentile(preds["latency_ms"], 95))
    result["latency_p95_ms"] = round(p95, 1)
    if p95 > THRESHOLDS["p95_latency_ms"]:
        alerts.append(f"Latensi p95 {p95:.0f} ms melebihi {THRESHOLDS['p95_latency_ms']} ms.")

    # 2) Data drift
    if len(preds) < THRESHOLDS["min_samples"]:
        alerts.append(f"Baru {len(preds)} prediksi (< {THRESHOLDS['min_samples']}): drift belum bisa dinilai andal.")
        drift = None
    else:
        reference = pd.read_csv(reference_path)
        live = pd.DataFrame(preprocess(list(preds["input"]), bundle), columns=features)
        drift = drift_report(reference, live, features)
        for _, r in drift[drift["status"] == "DRIFT"].iterrows():
            alerts.append(f"Drift pada '{r['fitur']}' (PSI={r['psi']}).")
    result["drift"] = drift

    # 3) Performa model (hanya untuk request yang sudah punya nilai aktual)
    perf = None
    if not fb.empty:
        m = preds.merge(fb[["request_id", "actual_value"]], on="request_id")
        if len(m) >= 30:
            mae = float(mean_absolute_error(m["actual_value"], m["prediction"]))
            r2 = float(r2_score(m["actual_value"], m["prediction"]))
            base = bundle["baseline"]
            perf = {"n": len(m), "MAE": round(mae, 1), "R2": round(r2, 3),
                    "MAE_baseline": round(base["MAE"], 1), "R2_baseline": round(base["R2"], 3)}
            if mae > base["MAE"] * (1 + THRESHOLDS["mae_degradation"]):
                alerts.append(f"MAE {mae:,.0f} naik {mae / base['MAE'] - 1:.0%} dari baseline {base['MAE']:,.0f}. "
                              "Pertimbangkan retraining.")
    result["performance"] = perf
    result["alerts"] = alerts

    if verbose:
        print(f"Prediksi: {result['n_predictions']} | umpan balik: {result['n_feedback']} | "
              f"latensi p95: {result['latency_p95_ms']} ms")
        if drift is not None:
            print("\nDrift fitur (PSI tertinggi di atas):")
            print(drift.to_string(index=False))
        if perf:
            print("\nPerforma:", perf)
        print("\nALERT:" if alerts else "\nTidak ada alert.")
        for a in alerts:
            print(" -", a)

    if save:
        out = {k: v for k, v in result.items() if k != "drift"}
        out["drift"] = None if drift is None else drift.to_dict("records")
        Path(log_dir).mkdir(parents=True, exist_ok=True)
        (Path(log_dir) / "monitoring_report.json").write_text(json.dumps(out, indent=2, default=str))
    return result


if __name__ == "__main__":
    run_monitoring()
```
